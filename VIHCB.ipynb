### Schritt 3: Beweismittel-Injection (Filtern & Einpflegen)
Dieses Skript filtert die Beweismittel strikt nach der Beweislogik: **1. Mangel -> 2. Akteur**. Anschließend werden die konkreten Referenzen und Zeitstempel in das Dokument injiziert, sodass jede Behauptung zwingend mit einem Nachweis gestützt wird.

```python
import pandas as pd
import os
import re

# 1. Laden des Master-Index (mit Semikolon als Trennzeichen, basierend auf der vorherigen Ausgabe)
try:
    df_emails = pd.read_csv('/content/EMAIL_MASTER_INDEX.csv', sep=';', on_bad_lines='skip')
except Exception as e:
    print(f"Fehler beim Laden des E-Mail-Index: {e}")
    df_emails = pd.DataFrame()

# 2. Definition der Filter-Logik (Mängel & Akteure)
mangel_keywords = ['Heizung', 'Wasser', 'Schimmel', 'Defekt', 'Mangel', 'Ausfall', 'Druck']
akteure_jobcenter = ['Jobcenter', 'JC']
akteure_verwaltung = ['Mutschall', 'Scheer', 'Pfaffenhofer', 'Verwaltung', 'Vermieter', 'Residea']

def filter_beweise(df, keywords):
    if df.empty:
        return df
    pattern = '|'.join(keywords)
    cols = ['Betreff', 'Kurzinhalt', 'Empfänger', 'Absender', 'Genannte Personen', 'Kategorie']
    available_cols = [c for c in cols if c in df.columns]
    if not available_cols:
        return df.head(0)
    # Suche in allen verfügbaren Textspalten
    mask = df[available_cols].apply(lambda x: x.astype(str).str.contains(pattern, case=False, na=False)).any(axis=1)
    return df[mask]

# 3. Beweise für das Jobcenter extrahieren (Fokus: Kosten, Regress, Ignoranz)
df_jc = filter_beweise(df_emails, akteure_jobcenter)
jc_beweise_text = ""
for idx, row in df_jc.head(8).iterrows():
    date = row.get('Datum', 'Unbekanntes Datum')
    subj = row.get('Betreff', 'Kein Betreff')
    beweis_id = row.get('Beweis-ID', f'E-JC-{idx}')
    jc_beweise_text += f"    *   **{date}** | Beweis-ID `{beweis_id}`: {subj}\n"

if not jc_beweise_text:
    jc_beweise_text = "    *   *Systemhinweis: Bitte EMAIL_MASTER_INDEX.csv auf vollständige Jobcenter-Korrespondenz prüfen.*\n"

# 4. Beweise für Verwaltung/Akteure extrahieren (Fokus: Mangel + Vorsatz durch Akteure)
df_vw = filter_beweise(df_emails, akteure_verwaltung)
df_vw_mangel = filter_beweise(df_vw, mangel_keywords)
vw_beweise_text = ""
for idx, row in df_vw_mangel.head(8).iterrows():
    date = row.get('Datum', 'Unbekanntes Datum')
    subj = row.get('Betreff', 'Kein Betreff')
    beweis_id = row.get('Beweis-ID', f'E-VW-{idx}')
    vw_beweise_text += f"    *   **{date}** | Beweis-ID `{beweis_id}`: {subj}\n"

# 5. Multimedia-Beweise (Fotos/Videos) zuordnen (Fokus: Heizung/Wasser/Zustand)
media_text = "\n    *   **Zusätzliche Medienbeweise (Zustandsdokumentation):**\n"
relevante_videos = [v for v in videos if 'Wasser' in v or 'Heizung' in v or 'Video' in v][:4]
for v in relevante_videos:
    media_text += f"        *   Video: `{os.path.basename(v)}`\n"

relevante_fotos = [f for f in fotos if 'Heizung' in f or 'Wasser' in f][:4]
if not relevante_fotos:
    relevante_fotos = fotos[:4] # Fallback
for f in relevante_fotos:
    media_text += f"        *   Foto: `{os.path.basename(f)}`\n"

# 6. Platzhalter im Dokument ersetzen (Robuster mit Regex)
file_path = '/content/BEWEISLOGIK_RECHTSWEG.md'
if os.path.exists(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        content = f.read()

    # Injektion der formatierten Beweise - re.sub ersetzt die ganze Zeile, in der der Platzhalter steht
    content = re.sub(r'.*\[PLATZHALTER_DOKUMENTE\].*', jc_beweise_text.rstrip('\n'), content)
    content = re.sub(r'.*\[PLATZHALTER_VIDEOS\].*', (vw_beweise_text + media_text).rstrip('\n'), content)

    with open(file_path, 'w', encoding='utf-8') as f:
        f.write(content)

    print("✅ Beweisakte wurde erfolgreich mit konkreten Nachweisen gestützt und aktualisiert!\n")
    from IPython.display import Markdown
    display(Markdown(content))
else:
    print(f"Fehler: Die Datei {file_path} wurde nicht gefunden.")
```

```python
import os
import re
import datetime
from IPython.display import display, Markdown

# Pfade zu den Dokumenten
anschreiben_path = '/content/ANSCHREIBEN_JOBCENTER.md'
cost_files = [
    '/content/05_REGRESS_FORDERUNGEN.md',
    '/content/08_Rechnungen_und_Einzelbetraege.md',
    '/content/07_Schadensausgaben_BK_WW_Heizkosten.md'
]

total_sum = 0.0
found_amounts = []

# 1. Beträge aus den Dokumenten extrahieren
print("Suche nach Beträgen in den Regress-Dokumenten...")
for file_path in cost_files:
    if os.path.exists(file_path):
        with open(file_path, 'r', encoding='utf-8') as f:
            content = f.read()
            # Suche nach Beträgen wie 1.234,56 € oder 1234,56 EUR
            amounts = re.findall(r'(\d{1,3}(?:\.\d{3})*,\d{2})\s*(?:€|EUR|Euro)', content)
            for amount in amounts:
                # Konvertiere deutsches Zahlenformat in Float
                val = float(amount.replace('.', '').replace(',', '.'))
                found_amounts.append(val)
                print(f"  - Gefunden in {os.path.basename(file_path)}: {amount} €")

# Gesamtsumme berechnen (wir summieren alle eindeutigen Beträge, um Doppelungen in Zusammenfassungen zu vermeiden)
# HINWEIS: Dies ist eine Schätzung. Wenn eine Gesamtsumme explizit im Text steht, wird sie evtl. mitaddiert.
if found_amounts:
    # Versuche eine intelligente Summierung (z.B. den größten Betrag nehmen, falls er die Summe der anderen ist)
    max_amount = max(found_amounts)
    sum_others = sum(found_amounts) - max_amount

    # Wenn der größte Betrag in etwa der Summe der anderen entspricht, war es vermutlich die Gesamtsumme im Dokument
    if abs(max_amount - sum_others) < 1.0:
        total_sum = max_amount
        print("\n-> Gesamtsumme wurde aus einer Zusammenfassung im Dokument erkannt.")
    else:
        total_sum = sum(found_amounts)
        print("\n-> Alle Einzelbeträge wurden summiert.")
else:
    print("Keine Beträge in den Standard-Dokumenten gefunden. Setze Platzhalter.")

today_date = datetime.datetime.today().strftime('%d.%m.%Y')
formatted_sum = f"{total_sum:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.') + " €"

# 2. Anschreiben aktualisieren
if os.path.exists(anschreiben_path):
    with open(anschreiben_path, 'r', encoding='utf-8') as f:
        letter_content = f.read()

    # Datum ersetzen
    letter_content = letter_content.replace('[Aktuelles Datum]', today_date)

    # Summe einfügen
    if "die Anerkennung der ausstehenden Regressforderungen" in letter_content and total_sum > 0:
        letter_content = letter_content.replace(
            "die Anerkennung der ausstehenden Regressforderungen",
            f"die Anerkennung der ausstehenden Regressforderungen in einer Gesamthöhe von **{formatted_sum}**"
        )

    with open(anschreiben_path, 'w', encoding='utf-8') as f:
        f.write(letter_content)

    display(Markdown(f"### ✅ Anschreiben aktualisiert\nDie Forderungssumme von **{formatted_sum}** wurde berechnet und zusammen mit dem Datum ({today_date}) eingefügt."))
    display(Markdown("---"))
    display(Markdown(letter_content))
    display(Markdown("---"))
else:
    print(f"❌ Datei {anschreiben_path} nicht gefunden.")
```

```python
import pandas as pd
import pdfkit
import os

# CSS-Styling für eine saubere PDF-Ausgabe
html_template = """
<!DOCTYPE html>
<html>
<head>
<meta charset="utf-8">
<style>
  body {{ font-family: Arial, sans-serif; font-size: 10px; }}
  h2 {{ text-align: center; }}
  table {{ border-collapse: collapse; width: 100%; page-break-inside: auto; }}
  tr {{ page-break-inside: avoid; page-break-after: auto; }}
  th, td {{ border: 1px solid #dddddd; text-align: left; padding: 6px; word-wrap: break-word; }}
  th {{ background-color: #f2f2f2; }}
</style>
</head>
<body>
  <h2>{title}</h2>
  {table}
</body>
</html>
"""

# Funktion zum Exportieren eines DataFrames als PDF
def export_df_to_pdf(df, title, filename):
    if df is not None and not df.empty:
        # Reduziere Spalten auf die wichtigsten für den Druck, um Platz zu sparen
        print_cols = ['Beweis-ID', 'Datum', 'Absender', 'Empfänger', 'Betreff', 'Kurzinhalt']
        available_cols = [c for c in print_cols if c in df.columns]
        df_print = df[available_cols]

        html_table = df_print.to_html(index=False, na_rep='', escape=False)
        final_html = html_template.format(title=title, table=html_table)

        pdf_path = f'/content/{filename}'
        try:
            pdfkit.from_string(final_html, pdf_path)
            print(f"✅ PDF erfolgreich erstellt: {pdf_path}")
        except Exception as e:
            print(f"❌ Fehler bei der PDF-Erstellung für {filename}: {e}")
    else:
        print(f"❌ DataFrame für {title} ist leer oder nicht vorhanden.")

# 1. Export Beweistabelle Verwaltung (für Staatsanwaltschaft)
if 'df_vw_mangel' in globals():
    export_df_to_pdf(df_vw_mangel.head(30), "Beweistabelle Verwaltung / Eigentümer", "Beweistabelle_Verwaltung.pdf")

# 2. Export Beweistabelle Jobcenter
if 'df_jc' in globals():
    export_df_to_pdf(df_jc.head(30), "Beweistabelle Jobcenter", "Beweistabelle_Jobcenter.pdf")
```

```python
import os
import re
import pandas as pd
import io
from IPython.display import display, Markdown

anlage_path = '/content/ANLAGE_REGRESS_DETAILAUFSTELLUNG.md'
cost_files = [
    '/content/05_REGRESS_FORDERUNGEN.md',
    '/content/08_Rechnungen_und_Einzelbetraege.md',
    '/content/07_Schadensausgaben_BK_WW_Heizkosten.md'
]

def extract_amounts_from_file(filepath):
    if not os.path.exists(filepath): return 0.0
    with open(filepath, 'r', encoding='utf-8') as f:
        content = f.read()
        amounts = re.findall(r'(\d{1,3}(?:\.\d{3})*,\d{2})\s*(?:€|EUR|Euro)', content)
        return sum([float(a.replace('.', '').replace(',', '.')) for a in amounts])

# 1. Originalsummen berechnen
original_sums = {}
for f in cost_files:
    original_sums[os.path.basename(f)] = extract_amounts_from_file(f)

# 2. Summen aus der Anlage parsen
if os.path.exists(anlage_path):
    with open(anlage_path, 'r', encoding='utf-8') as f:
        content = f.read()

    # Tabelle extrahieren (ohne Header-Trenner)
    table_lines = [line for line in content.split('\n') if '|' in line and '---' not in line]

    if len(table_lines) > 1:
        # Mittels pandas als CSV (mit '|' als Trennzeichen) einlesen
        csv_data = '\n'.join(table_lines)
        df_anlage = pd.read_csv(io.StringIO(csv_data), sep='|').iloc[:, 1:-1]
        df_anlage.columns = [c.strip() for c in df_anlage.columns]

        # Beträge in Float umwandeln
        df_anlage['Betrag_float'] = df_anlage['Betrag'].str.extract(r'(\d{1,3}(?:\.\d{3})*,\d{2})')[0].str.replace('.', '').str.replace(',', '.').astype(float)

        # Nach Quelldokument gruppieren und summieren
        df_grouped = df_anlage.groupby('Quelldokument')['Betrag_float'].sum().reset_index()

        display(Markdown("### 📊 Abgleich der Rechnungssummen (Anlage vs. Originale)"))

        all_match = True
        for idx, row in df_grouped.iterrows():
            doc = str(row['Quelldokument']).strip()
            sum_anlage = row['Betrag_float']
            sum_orig = original_sums.get(doc, 0.0)

            status = "✅ Stimmt exakt überein" if abs(sum_anlage - sum_orig) < 0.01 else "❌ Abweichung festgestellt"
            if "❌" in status:
                all_match = False

            display(Markdown(f"**Dokument:** `{doc}` \n* **Summe in der Anlage:** {sum_anlage:,.2f} € \n* **Summe im Original:** {sum_orig:,.2f} € \n* **Ergebnis:** {status}"))
            display(Markdown("---"))

        if all_match:
            display(Markdown("**Fazit:** Alle Beträge der Detailaufstellung stimmen exakt mit den Originaldokumenten überein."))
        else:
            display(Markdown("**Fazit:** Es gibt Diskrepanzen! Bitte die Extraktion überprüfen."))
else:
    print(f"❌ Datei {anlage_path} nicht gefunden.")
```

```python
import zipfile
import os
from IPython.display import display, Markdown

zip_path = '/content/Beweisakte_Uebergabe.zip'
logic_path = '/content/BEWEISLOGIK_RECHTSWEG.md'

if os.path.exists(zip_path) and os.path.exists(logic_path):
    with open(logic_path, 'r', encoding='utf-8') as f:
        logic_content = f.read()

    with zipfile.ZipFile(zip_path, 'r') as zipf:
        zip_files = zipf.namelist()

    core_docs = [
        'BEWEISLOGIK_RECHTSWEG.md',
        'ANSCHREIBEN_JOBCENTER.md',
        'ANLAGE_REGRESS_DETAILAUFSTELLUNG.md'
    ]

    linked_files = []
    unlinked_files = []

    for file in zip_files:
        if file in core_docs:
            continue

        basename = os.path.basename(file)
        name_without_ext = os.path.splitext(basename)[0]

        # Prüfe, ob der Dateiname oder die ID (z.B. E-0005) im Text vorkommt
        if basename in logic_content or name_without_ext in logic_content:
            linked_files.append(basename)
        else:
            unlinked_files.append(basename)

    output = "### 🔍 Cross-Check: ZIP-Archiv vs. Beweislogik\n"

    output += "#### ✅ Erfolgreich in der Beweislogik referenzierte Dateien:\n"
    for f in linked_files:
        output += f"- `{f}`\n"
    if not linked_files:
        output += "- *Keine*\n"

    output += "\n#### ⚠️ Im ZIP enthalten, aber nicht namentlich in der Beweislogik erwähnt:\n"
    for f in unlinked_files:
        output += f"- `{f}`\n"
    if not unlinked_files:
        output += "- *Alle Beweismittel sind einwandfrei verknüpft!*\n"

    output += "\n*(Hinweis: Tabellen wie 'Beweistabelle_Jobcenter.xlsx' gelten oft als Metadokumente und müssen nicht zwingend im Haupttext stehen, solange sie im Anschreiben referenziert sind.)*"

    display(Markdown(output))
else:
    print("❌ ZIP-Archiv oder Beweislogik-Dokument fehlt.")
```
